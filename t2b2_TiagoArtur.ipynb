```python
# Consulta 1a)

# Para listar os clientes da Região Norte, a gente pensou assim:
# a gente sabe que os estados do Norte são AC, AM, AP, PA, RO, RR e TO.
# Então, a gente vai na coleção de clientes e pede pra ela achar todo mundo
# onde o 'uf' do 'endereco' esteja dentro dessa nossa lista de estados.
# No final, a gente mostra só o nome do cliente, a cidade e o estado,
# pra ficar limpo e fácil de ler.

print("\n--- Consulta 1a) Clientes da Região Norte ---")
regiao_norte_ufs = ["AC", "AM", "AP", "PA", "RO", "RR", "TO"]

try:
    clientes_norte = clientes_collection.find(
        {"endereco.uf": {"$in": regiao_norte_ufs}},
        {"nome": 1, "endereco.cidade": 1, "endereco.uf": 1, "_id": 0}
    )

    if clientes_norte:
        count_norte = 0
        for cliente in clientes_norte:
            print(f"- Nome: {cliente['nome']}, Cidade: {cliente['endereco']['cidade']}, UF: {cliente['endereco']['uf']}")
            count_norte += 1
        print(f"Total de clientes na Região Norte: {count_norte}")
    else:
        print("Nenhum cliente encontrado na Região Norte.")
except Exception as e:
    print(f"Erro na consulta de clientes da Região Norte: {e}")
```

```python
# Consulta 1b)


# Para esse desenvolvimento, buscamos o pedido do cliente e a categoria que está no nosso catálogo de produtos e, então,
# usamos o aggregate, que é a ferramenta do MongoDB pra fazer análises mais complexas.
# A lógica é:
# 1. A gente 'une' os clientes com os produtos usando o `$lookup`. A gente 'cola' os detalhes do produto
#    no documento do cliente, usando aquela referência `codProduto_ref` que criamos.
# 2. Depois, o `$unwind` 'abre' os detalhes do produto que vieram no join, pra gente acessar direto.
# 3. `$group`: a gente agrupa todo mundo pela 'categoria.nome' do produto
#    e usa o `$sum` pra somar a 'quantidade' de cada pedido dentro daquela categoria.
# 4. Pra deixar bonito, a gente `$sort` os resultados pela quantidade total, do maior pro menor,
#    e no `$project` a gente só mostra a categoria e a quantidade somada.

print("\n--- Consulta 1b) Quantidade Total de Produtos por Categoria ---")
try:
    produtos_por_categoria = clientes_collection.aggregate([
        {
            "$lookup": {
                "from": "produtos",
                "localField": "pedido.codProduto_ref",
                "foreignField": "codProduto",
                "as": "detalhes_produto"
            }
        },
        {
            "$unwind": "$detalhes_produto"
        },
        {
            "$group": {
                "_id": "$detalhes_produto.categoria.nome",
                "quantidadeTotalVendida": {"$sum": "$pedido.quantidade"}
            }
        },
        {
            "$sort": {"quantidadeTotalVendida": -1}
        },
        {
            "$project": {
                "_id": 0,
                "Categoria": "$_id",
                "QuantidadeTotalVendida": "$quantidadeTotalVendida"
            }
        }
    ])

    print("Quantidade de produtos vendidos por categoria:")
    found_categories = False
    for item in produtos_por_categoria:
        print(f"- Categoria: {item['Categoria']}, Total Vendido: {item['QuantidadeTotalVendida']}")
        found_categories = True
    if not found_categories:
        print("Nenhuma categoria de produto encontrada ou dados insuficientes.")
except Exception as e:
    print(f"Erro na consulta de produtos por categoria: {e}")
```

```python
# Consulta 1c)

# Pra essa consulta, a ideia é descobrir quais pedidos renderam mais que mil.
# Primeiro, calculamos o total de cada pedido multiplicando a 'quantidade' pelo 'valor_unitario' do pedido.
# A lógica é:
# 1. Usamos `$addFields` pra criar um campo novo chamado 'valor_total_pedido', onde a gente guarda
#    o resultado da multiplicação da quantidade pelo valor unitário de cada item do pedido.
# 2. Depois, a gente usa `$match` pra filtrar somente os pedidos onde esse 'valor_total_pedido' é maior que 1000.
# 3. E no `$project`, a gente escolhe quais campos quer exibir pra não poluir a tela,
#    mostrando o nome do cliente, o número do pedido, o produto e o valor total que a gente calculou.

print("\n--- Consulta 1c) Pedidos com Valor Total Acima de R$ 1000,00 ---")
try:
    pedidos_acima_mil = clientes_collection.aggregate([
        {
            "$addFields": {
                "valor_total_pedido": {"$multiply": ["$pedido.quantidade", "$pedido.valor_unitario"]}
            }
        },
        {
            "$match": {
                "valor_total_pedido": {"$gt": 1000}
            }
        },
        {
            "$project": {
                "_id": 0,
                "Cliente": "$nome",
                "NumPedido": "$pedido.num_pedido",
                "ProdutoComprado": "$pedido.produto",
                "Quantidade": "$pedido.quantidade",
                "ValorUnitarioNaCompra": "$pedido.valor_unitario",
                "ValorTotalPedido": "$valor_total_pedido"
            }
        },
        {
            "$sort": {"ValorTotalPedido": -1}
        }
    ])

    print("Pedidos com valor total superior a R$ 1000,00:")
    found_orders = False
    for pedido in pedidos_acima_mil:
        print(f"- Cliente: {pedido['Cliente']}, Pedido Nº: {pedido['NumPedido']}, Produto: {pedido['ProdutoComprado']}, Total: R$ {pedido['ValorTotalPedido']:.2f}")
        found_orders = True
    if not found_orders:
        print("Nenhum pedido encontrado com valor total acima de R$ 1000,00.")
except Exception as e:
    print(f"Erro na consulta de pedidos acima de R$ 1000,00: {e}")
```

```python
# Consulta 1d)

# Nesta consulta, para conseguirmos achar os 5 produtos que mais saíram
# usamos aggregate de novo, porque a quantidade está no pedido
# e o nome do produto que a gente quer mostrar está na coleção 'produtos'.
# A lógica é:
# 1. A gente filtra os pedidos pra ter certeza que tem a referência do produto (`codProduto_ref`).
# 2. Aplicamos `$group` nos pedidos pelo `codProduto_ref` (que é o ID do produto) e soma a 'quantidade' de cada um.
# 3. Depois, a gente `$sort` essa soma em ordem decrescente pra ver os maiores primeiro.
# 4. Usamos `$limit 5` pra pegar só os cinco primeiros.
# 5. E pra finalizar, outro `$lookup` pra 'colar' os detalhes do produto (tipo o nome e a categoria)
#    a cada um desses top 5 que achamos, para mostrar adequadamente no `$project`.

print("\n--- Consulta 1d) Cinco Produtos Mais Vendidos ---")
try:
    cinco_mais_vendidos = clientes_collection.aggregate([
        {
            "$match": {
                "pedido.codProduto_ref": {"$exists": True}
            }
        },
        {
            "$group": {
                "_id": "$pedido.codProduto_ref",
                "quantidadeTotalVendida": {"$sum": "$pedido.quantidade"}
            }
        },
        {
            "$sort": {"quantidadeTotalVendida": -1}
        },
        {
            "$limit": 5
        },
        {
            "$lookup": {
                "from": "produtos",
                "localField": "_id",
                "foreignField": "codProduto",
                "as": "detalhes_produto"
            }
        },
        {
            "$unwind": "$detalhes_produto"
        },
        {
            "$project": {
                "_id": 0,
                "Produto": "$detalhes_produto.nome",
                "Categoria": "$detalhes_produto.categoria.nome",
                "QuantidadeTotalVendida": "$quantidadeTotalVendida"
            }
        }
    ])

    print("Os 5 produtos mais vendidos:")
    found_top_products = False
    for produto in cinco_mais_vendidos:
        print(f"- Produto: {produto['Produto']} (Categoria: {produto['Categoria']}), Total Vendido: {produto['QuantidadeTotalVendida']}")
        found_top_products = True
    if not found_top_products:
        print("Nenhum produto encontrado na lista dos 5 mais vendidos.")
except Exception as e:
    print(f"Erro na consulta dos 5 produtos mais vendidos: {e}")
```

```python
# Consulta 2a) Resolvemos listar os clientes Pessoa Jurídica (tipo "J") em Porto Alegre:

# Aqui a gente quer saber quais empresas (Pessoa Jurídica) da nossa base são de Porto Alegre.
# A gente usa o `find()` de novo, porque é uma consulta mais simples de filtragem.
# A lógica é:
# A gente busca na coleção de clientes documentos que tenham o 'tipo' igual a "J" E o 'endereco.cidade' igual a "Porto Alegre".
# No final, a gente mostra o nome fantasia, a razão social, o CNPJ, e ordena por nome.

print("\n--- Consulta 2a) Clientes Pessoa Jurídica em Porto Alegre ---")
try:
    clientes_pj_poa = clientes_collection.find(
        {"tipo": "J", "endereco.cidade": "Porto Alegre"},
        {"nome": 1, "cnpj": 1, "razao_social": 1, "endereco.cidade": 1, "_id": 0}
    ).sort("nome", 1)

    print("Clientes Pessoa Jurídica em Porto Alegre:")
    found_pj_poa = False
    for cliente in clientes_pj_poa:
        print(f"- Nome Fantasia: {cliente['nome']}, Razão Social: {cliente.get('razao_social', 'N/A')}, CNPJ: {cliente.get('cnpj', 'N/A')}")
        found_pj_poa = True
    if not found_pj_poa:
        print("Nenhum cliente Pessoa Jurídica encontrado em Porto Alegre.")
except Exception as e:
    print(f"Erro na consulta de clientes PJ em Porto Alegre: {e}")
```

```python
# Consulta 2b) Nessa resolvemos calcular o valor total de vendas por UF (estado):

# Pra essa consulta, a gente quer ter uma ideia de qual estado está gerando mais receita para o nosso comércio.
# Usamos o Aggregation Framework de novo, porque precisamos somar valores em grupos.
# A lógica é:
# 1. A gente usa o `$group` para agrupar todos os pedidos pelo 'endereco.uf' (o estado do cliente).
# 2. Dentro de cada grupo (cada estado), a gente usa o `$sum` pra somar o valor total de CADA pedido
#    (que é a 'quantidade' multiplicada pelo 'valor_unitario').
# 3. No final, a gente `$sort` os resultados pra ver os estados com maior volume de vendas primeiro,
#    e no `$project` a gente só mostra o estado (UF) e o 'ValorTotalVendas'.

print("\n--- Consulta 2b) Valor Total de Vendas por UF ---")
try:
    vendas_por_uf = clientes_collection.aggregate([
        {
            "$group": {
                "_id": "$endereco.uf",
                "valorTotalVendas": {"$sum": {"$multiply": ["$pedido.quantidade", "$pedido.valor_unitario"]}}
            }
        },
        {
            "$sort": {"valorTotalVendas": -1}
        },
        {
            "$project": {
                "_id": 0,
                "UF": "$_id",
                "ValorTotalVendas": "$valorTotalVendas"
            }
        }
    ])

    print("Valor total de vendas por Unidade Federativa (UF):")
    found_sales_uf = False
    for uf_venda in vendas_por_uf:
        print(f"- UF: {uf_venda['UF']}, Valor Total de Vendas: R$ {uf_venda['ValorTotalVendas']:.2f}")
        found_sales_uf = True
    if not found_sales_uf:
        print("Nenhum registro de vendas encontrado por UF.")
except Exception as e:
    print(f"Putz, deu erro na consulta de vendas por UF: {e}")

print("\n--- Todas as consultas foram implementadas com sucesso! ---")
```

#Justificativas



Justificativas:
 -Para começar, escolhemos apenas o Cliente como coleção principal, pois, além de ser o centro das operações (com pedidos, endereço e telefone), fazia mais sentido começar modelando por ele. Depois de um tempo de trabalho, percebemos que, por estarmos usando a versão gratuita da ferramenta, não conseguíamos expandir muito a modelagem. Porém, no MongoDB, foi possível criar a coleção produtos e vincular eles com os clientes usando referências em MongoPy, sendo possível resolver a consulta b com sucesso.

 -Colocamos documentos dentro de outros (por exemplo, endereço dentro de cliente, ou itens dentro de pedido), porque isso deixa a estrutura mais próxima de como usamos os dados no mundo real. Assim, dá pra consultar tudo de uma vez só, sem precisar ficar "juntando" dados de várias coleções, como seria no modelo não relacional tradicional de um sistema de comércio.

 -Ao incluir os pedidos e os produtos diretamente no documento do cliente, pensamos em melhorar o desempenho na leitura de dados. Assim, podemos mostrar rapidamente todas as informações de um cliente com seus pedidos, sem fazer várias buscas.
  Essa separação foi pensada pra deixar claro os diferentes tipos de cliente, cada um com seus próprios campos obrigatórios. Isso ajuda na organização e também evita erros na hora de cadastrar.

-Por fim, definimos tipos como string, integer, boolean, e marcamos campos obrigatórios pra garantir que os dados sempre tenham o formato esperado.
